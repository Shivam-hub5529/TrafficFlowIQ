# 02 Preprocessing, labels, splits
Shows the leakage controls.

```
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from trafficflowiq.data import load_panel
panel = load_panel('synthetic')  # use 'kaggle' after filling the column map
```

```
from trafficflowiq.features import build_queue_features, valid_rows
from trafficflowiq.labels import build_labels
from trafficflowiq.splits import split_boundaries, split_masks
feats = build_queue_features(panel); labels = build_labels(panel)
df = feats.join(labels); df = df[valid_rows(df) & df.y_any.notna()]
vs, ts = split_boundaries(panel.index); print('val starts', vs, 'test starts', ts)
```

```
m = split_masks(df.index.get_level_values('timestamp'), vs, ts)
for k,v in m.items(): print(k, int(v.sum()), 'rows, queue rate', round(df[v].y_any.mean(),3))
```
