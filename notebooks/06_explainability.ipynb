# 06 Explainability
Global importance and local explanations. LightGBM `pred_contrib=True` gives exact tree-SHAP; the optional `shap` package adds plots.

```
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from trafficflowiq.data import load_panel
panel = load_panel('synthetic')  # use 'kaggle' after filling the column map
```

```
from trafficflowiq.modelio import load_bundle
from trafficflowiq.features import build_queue_features
from trafficflowiq.explain import explain
from pathlib import Path
objs, meta = load_bundle(Path('../models/queue/v1.0.0'), ['queue_model'])
feats = build_queue_features(panel)
row = feats.xs(3, level='link_id').loc[['2025-02-12 16:25:00']]
c = objs['queue_model'].predict(row[meta['features']], pred_contrib=True)[0][:-1]
explain(dict(zip(meta['features'], c)), row.iloc[0].to_dict())
```

```
imp = pd.Series(objs['queue_model'].booster_.feature_importance('gain'), index=meta['features']).sort_values()
imp.tail(15).plot.barh(figsize=(6,5), title='Top features by gain'); plt.show()
```
