# 01 EDA
Questions: how often are links queued, when, where, and how much data is missing (especially ramps)?

```
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from trafficflowiq.data import load_panel
panel = load_panel('synthetic')  # use 'kaggle' after filling the column map
```

```
ffs = panel.meta['ffs_kmh']
ratio = panel.speed.div(ffs, axis=1)
q = (ratio<=0.6).astype(float).where(ratio.notna())
print('Queue prevalence by link:'); print(q.mean().round(3))
```

```
q.groupby(q.index.hour).mean().plot(figsize=(9,3), title='Queued share by hour of day'); plt.show()
```

```
print('Missing share:'); print(pd.DataFrame({'speed':panel.speed.isna().mean(),'flow':panel.flow.isna().mean(),'ramp_bad':panel.ramp_bad.mean()}).round(3))
```
