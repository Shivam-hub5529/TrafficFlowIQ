# 04 LightGBM queue model
Also the train/serve skew check: the risk printed here must equal the API's risk for the same link and time.

```
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from trafficflowiq.data import load_panel
panel = load_panel('synthetic')  # use 'kaggle' after filling the column map
```

```
from trafficflowiq.modelio import load_bundle
from trafficflowiq.features import build_queue_features
from pathlib import Path
objs, meta = load_bundle(Path('../models/queue/v1.0.0'), ['queue_model','queue_calibrator'])
feats = build_queue_features(panel)
row = feats.xs(3, level='link_id').loc[['2025-02-12 16:25:00']][meta['features']]
raw = objs['queue_model'].predict_proba(row)[:,1]
print('risk =', objs['queue_calibrator'].predict(raw))
```
